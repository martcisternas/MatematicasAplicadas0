* Pregunta 1

El tipo de funcion que permite calcular el tiempo de encriptacion es Lineal, ya que va aumentando de manera constante dependiendo de la cantidad de archivos.

La forma algebraica de la funcion sería:   a(x) = 0,67x - 12 

```python
def a(x):
    return 0.67 * x - 12

print(f'Cuando el tamaño del archivo es de 62MB el tiempo con el metodo "A" es de {a(62)} segundos.')
```

```python
def g(x):
    return 0.01 * x **2 - 0.4 * x + 8

print(f'Cuando el tamaño del archivo es de 62MB el tiempo con el metodo "B" es de {g(62):.2f} segundos.')
```

* Pregunta 2

El metodo que empleo menos tiempo en la encriptacion de 62 MB de tamaño del archivo fue el metodo "B" con "21.64" segundos.

```python
import numpy as np
import matplotlib.pyplot as plt

def a(x):
    return 0.67 * x - 12

def g(x):
    return 0.01 * x **2 - 0.4 * x + 8

x = np.arange(20, 101, 10)
plt.plot(x, a(x), label = 'Metodo A')
plt.plot(x, g(x), label = 'Metodo B')
plt.title('Metodos de encriptacion en MB.')
plt.xlabel('Tamaño del archivo en MB.')
plt.ylabel('Tiempo en segundos.')
plt.grid(True)
plt.yticks(np.arange(0,70.1,10))
plt.xticks(np.arange(20,101,10))
plt.legend()
plt.show()
```

```python
import numpy as np
from scipy.optimize import fsolve

def a(x):
    return 0.67 * x - 12

def g(x):
    return 0.01 * x **2 - 0.4 * x + 8

def interseccion(x):
    return a(x)- g(x)

x=np.linspace(20,30,1)
solucion=fsolve(interseccion,x)

print(f'Primer intersecto a los = {solucion[0]:.2f} MB por segundo')

x=np.linspace(70,90,1)
solucion1=fsolve(interseccion,x)

print(f'Segundo intersecto a los = {solucion1[0]:.2f} MB por segundo')

print(" ")

def a(x):
    return 0.67 * x - 12

print(f'Tiempo primer intersecto en {a(solucion[0]):.2f} segundos.')

def g(x):
    return 0.01 * x **2 - 0.4 * x + 8

print(f'Tiempo segundo intersecto en {a(solucion1[0]):.2f} segundos.')

```
