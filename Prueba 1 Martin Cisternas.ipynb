```python
import numpy as np
import matplotlib.pyplot as plt

def f(x):
    return 320 * x

def g(x):
    return 280 * x + 160

x = np.arange(0, 11.1, 5)
plt.plot(x, f(x), label = 'Sitio 1')
plt.plot(x, g(x), label = 'Sitio 2')
plt.title('Valores de tablet en base a cantidad.')
plt.xlabel('Cantidad tablet comprados.')
plt.ylabel('Valor en dolares.')
plt.xticks(np.arange(0,10.1,1))
plt.yticks(np.arange(0,3301,300))
plt.grid(True)
plt.legend()
plt.show()
```
