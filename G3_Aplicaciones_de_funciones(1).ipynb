# Problema 5

Te contrataron como asesor en una campaña de marketing digital y necesitas equilibrar el ingreso por ventas con los costos de inversión.

Las funciones de ingresos por ventas $I(x)$ y costos de inversión $C(x)$ (ambas en miles de pesos) son:

<br>

$$I(x) = - 20x + 1.000$$

$$C(x) = 200x + 500$$

<br>

Donde $x$ corresponde al tiempo transcurrido desde que se inició la campaña (en semanas).

<br>

1. Analice el comportamiento del ingreso por ventas y del costo de inversión a medida que aumenta el tiempo desde que se inició la campaña. Considere la pendiente y el intercepto de cada función para realizar el análisis.
2. Grafique ambas funciones en un mismo gráfico, indicando nombre
de los ejes. Utilice *Matplotlib*.
3. Determine, utilizando *Python*, el punto de equilibrio entre ambas funciones e interprete el resultado.
4. Utilizando el gráfico y lo realizado en el punto $3$, ¿qué puedes concluir? ¿Aconsejarías mantener la campaña transcurridas $5$ semanas desde el inicio de la campaña?

*Pregunta 1*
* Interpretación $I(x)$: La pendiente indica que, por cada semana transcurrida, el ingreso decrece 20 mil pesos. Mientras que el intercepto indica que la ganancia inicial (o base) es de 1.000.000 pesos.
* Interpretación $C(x)$: La pendiente indica que, por cada semana transcurrida, el costo aumenta 200 mil pesos.Mientras que el intercepto indica que el costo base es de 500.000 pesos.

```python
#Pregunta 2
import numpy as np
import matplotlib.pyplot as plt
def I(x):
    return -20*x + 1000
def C(x):
    return 200*x + 500
x = np.arange(0,6.1,0.1)
plt.plot(x, I(x), label = 'Ingreso')
plt.plot(x, C(x), label = 'Costo')
plt.title('Costo e ingreso en función del tiempo')
plt.xlabel('Tiempo (semanas)')
plt.ylabel('Costo o ingreso (miles de pesos)')
plt.grid(True)
plt.legend()
plt.show()
```

```python
#Pregunta 3
import numpy as np
from scipy.optimize import fsolve
def I(x):
    return -20*x + 1000
def C(x):
    return 200*x + 500
#definir una nueva función f(x)-g(x)
def interseccion(x):
    return I(x)-C(x)
x=np.linspace(0,3,1)
solucion=fsolve(interseccion,x)
y_int = I(solucion[0])
print(f'El punto intersección es x={solucion[0]:.2f}, y = {y_int:.2f}')
```

* Interpretación: El ingreso y el costo es el mismo transcurridas 2,27 semanas, con un valor de $954.55 mil pesos. Luego de ese tiempo, el costo es mayor al ingreso.

# Problema 6

Eres responsable de una granja de servidores, y necesitas optimizar la energía utilizada $E(x)$ (en kWh), en función de la cantidad de servidores activos $x$. Luego de varias mediciones y tratamiento de datos encuentras que la función que modela la relación entre ambas variables es:

$$E(x) = 200 \sqrt x + 500$$

<br>

1. Define variable dependiente e independiente, indicando unidad de medida.
2. Grafica la función utilizando *Matplotlib*.
3. Determina la energía utilizada cuando hay $25$ servidores activos.
4. Encuentra el número máximo de servidores activos para que la energía no supere los $1.900$ kWh. Utiliza *Python* para calcular.

* Variable dependiente: Energía (kWh)
* Variable independiente: N° servidores (unidades)

```python
import numpy as np
import matplotlib.pyplot as plt
def E(x):
    return 200 * np.sqrt(x) + 500
x = np.arange(0,25.1,0.1)
plt.plot(x, E(x), label = 'E(x)')
plt.title('Energía en función del N° servidores')
plt.xlabel('N° servidores (unidades)')
plt.ylabel('Consumo energía (kWh)')
plt.grid(True)
plt.legend()
plt.show()
print(f'3) Con 25 servidores, se consumen {E(25)} kWh')
```

```python
import numpy as np
from scipy.optimize import fsolve
def E(x):
    return 200 * np.sqrt(x) + 500
#Definir la nueva función igualando a cero
def f(x):
    return E(x)-1900
#Valor/es inicial/es de la aproximación
x = np.linspace(0,10,1)
solucion = fsolve(f, x)
print(f'Se consumen 1900 kWh con {solucion[0]:.0f} de servidores')
```

Las funciones de ingresos por suscriptores $I(x)$ y costos de operación $C(x)$ (ambos en miles de pesos) son:

<br>

$$I(x) = 10x^{2} + 50x$$

$$C(x) = 5x^{2} + 80x + 100$$

<br>

Donde $x$ corresponde al número de suscriptores (en miles).

<br>

1. Debes determinar dónde se equilibran los ingresos por suscripciones con los costos de operación.
2. Analice en qué tramo los costos de operación son mayores que los ingresos por suscriptores.

```python
import numpy as np
import matplotlib.pyplot as plt
def I(x):
    return 10*x**2 + 50*x
def C(x):
    return 5*x**2 + 80*x + 100
x = np.arange(0,12.1,0.1)
plt.plot(x, I(x), label = 'Ingreso')
plt.plot(x, C(x), label = 'Costo')
plt.title('Costo e ingreso en función del n° suscriptores')
plt.xlabel('N° suscriptores (miles)')
plt.ylabel('Costo o ingreso (miles de pesos)')
plt.grid(True)
plt.legend()
plt.show()
```

```python
import numpy as np
from scipy.optimize import fsolve
def I(x):
    return 10*x**2 + 50*x
def C(x):
    return 5*x**2 + 80*x + 100
#definir una nueva función f(x)-g(x)
def interseccion(x):
    return I(x)-C(x)
x=np.linspace(8,10,1)
solucion=fsolve(interseccion,x)
print(f'1) Los ingresos y costos se equilibran luego de {solucion[0]:.2f} mil suscriptores')
```
