## Ejercicio 1: Rendimiento de Servidores en la Nube (Sucesión por Tramos)

Una empresa de infraestructura en la nube (*Cloud Services*) está analizando el comportamiento del tráfico de sus servidores durante el lanzamiento de un nuevo videojuego multijugador. El número de peticiones por segundo $a_n$ en el minuto $n$ del evento se define mediante la siguiente sucesión por tramos:

$$a_n = \begin{cases}
3n^2 + 10 & \text{si } n \le 10 \\[6pt]
50n - 100 & \text{si } n > 10
\end{cases}$$

**Responda a las siguientes cuestiones:**

1. Obtenga los valores de las peticiones por segundo para los minutos $n = 5$, $n = 10$, $n = 11$ y $n = 15$.
2. Calcule el total acumulado de peticiones recibidas desde el inicio hasta el minuto 20
3. Calcule el total acumulado de peticiones recibidas desde el minuto 7 hasta el minuto 12 .

## Ejercicio 2: Puntos de Experiencia en un RPG 

En un juego de rol (RPG) desarrollado por un estudio independiente, los puntos de experiencia (XP) requeridos para pasar del nivel $n-1$ al nivel $n$ siguen una progresión aritmética. Se sabe que para subir al nivel 2 ($n=2$) se necesitan $450$ XP, y para subir al nivel 6 ($n=6$) se necesitan $1.050$ XP.

**Responda a las siguientes cuestiones:**

1. Determine la regla general $a_n$ para la cantidad de XP necesaria para alcanzar el nivel $n$.
2. Calcule la cantidad de XP requerida para alcanzar los niveles $n = 12$ y $n = 25$.
3. Calcule la cantidad total acumulada de XP que un jugador necesita ganar para avanzar desde el nivel 10 hasta alcanzar el nivel 20.



## Ejercicio 3: Expansión de Almacenamiento en Data Center (Sucesión Geométrica)

Una corporación de tecnología de datos proyecta el crecimiento de su capacidad de almacenamiento en petabytes (PB) mes a mes. En el mes 1 ($n=1$), la capacidad agregada fue de $16$ PB, y en el mes 4 ($n=4$), la capacidad agregada fue de $128$ PB. El crecimiento mensual del almacenamiento sigue una progresión geométrica.

**Responda a las siguientes cuestiones:**

1. Encuentre la razón $r$ de la sucesión y la expresión algebraica general $g_n$ para la capacidad agregada en el mes $n$.
2. Obtenga los valores de capacidad agregada correspondientes a los meses $n = 5$ y $n = 8$.
3. Determine el total de almacenamiento en petabytes agregados a la red desde el mes 5 hasta el mes 12, ambos inclusive.

## Ejercicio 4: Comparativa de Algoritmos de Búsqueda (Aritmética vs. Geométrica)

Un laboratorio de inteligencia artificial está evaluando dos algoritmos para procesar grandes volúmenes de datos en servidores de alto rendimiento:

* **Algoritmo A (Crecimiento Lineal):** El tiempo de procesamiento en milisegundos $A_n$ para un lote de tamaño $n$ sigue una **progresión aritmética**. Se sabe que para un lote $n=2$ tarda $120$ ms y para un lote $n=7$ tarda $320$ ms.
* **Algoritmo B (Crecimiento Exponencial):** El tiempo de procesamiento en milisegundos $B_n$ para el mismo lote $n$ sigue una **progresión geométrica**. Se sabe que para un lote $n=2$ tarda $20$ ms y para un lote $n=6$ tarda $320$ ms.

**Responda a las siguientes cuestiones:**

1. Obtenga la expresión general para cada uno de los algoritmos: $A_n$ (encontrando el primer término $a_1$ y la diferencia $d$) y $B_n$ (encontrando el primer término $b_1$ y la razón $r$).
2. Calcule el tiempo de procesamiento que requiere cada algoritmo para los lotes $n = 5$ y $n = 8$. ¿Cuál de los dos algoritmos es más eficiente (tarda menos tiempo) para cada uno de esos lotes?
3. Calcule la suma total del tiempo acumulado al procesar de forma secuencial los lotes desde $n = 3$ hasta $n = 7$, ambos inclusive, para ambos algoritmos. Determine cuál algoritmo acumula menor tiempo total en ese rango.
