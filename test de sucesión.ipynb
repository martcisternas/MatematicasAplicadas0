Sucesión: $ h_i = 300000 \cdot 1.02^{i-1} $

```python
def h(i):
    return 300000 * 1.02 **(i-1)

print(h(1))
print(h(5))
```

* Listar los primeros 5 términos de la sucesión.


```python
for i in range(1,6):
    print(f'Término {i}={h(i)}')
```

Almacenar en una lista los primeros 10 térmions, y presentar en pantalla los primeros 5.

```python
lista = []

for i in range (10):
    print(f'Término {i} = {h(i)}')
    lista.append(h(i+1))

print()

print(lista) #Se puede hacer rapido asi lista[0:4]

print()

for i in range(1,6):
    print(f'Término {i} = {lista[i-1]}')
```
