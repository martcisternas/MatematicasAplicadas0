# Problema 2

Considere la sucesión $g_{n} = 5n^{3}$. Implemente un código en *Python* que muestre en pantalla:

3. Determinar si el término $40.000$ pertenece a la sucesión. Si es así, indicar en qué posición se encuentra.

```python
def g(n):
    return 5*n**3
for i in range (1,5):
    print(f'Término {i} = {g(i)}')
print('=> Pregunta 2')
for i in range (9,13):
    print(f'Término {i} = {g(i)}')
```

```python
def g(n):
    return 5*n**3
valor = 40000
dif = valor
i=0
while dif > 0:
    i+=1
    dif = valor - g(i)
if dif ==0:
    print(f'El término pertenece y ocupa la posición {i}')
else: 
    print('El término NO pertenece a la sucesión')
```

# Problema 11

Una aplicación de redes sociales se estrenó el $1$ de enero con una cierta cantidad de usuarios. A partir de ahí, el número de usuarios nuevos de cada día fue siempre el mismo. El $3$ de enero tenía $500$ usuarios en total y el $10$ de enero tenía $1.501$ usuarios totales.Determine:

1. ¿Con cuántos usuarios comenzó a funcionar la aplicación?
2. ¿Cuántos usuarios tendrá la aplicación el $31$ de enero?
3. Si la aplicación tiene $5.505$ usuarios, ¿cuántos meses lleva funcionando?



* Es una sucesión aritmética, ya que el n° de usuarios nuevos diarios siempre fue el mismo. 
$$a_3 = 500$$
$$a_{10} = 1501$$
Falta por obtener $d$ y $a_1$
* Para obtener $d$
$$a_3 + 7d = a_{10}$$
$$ d = (a_{10}-a_3)/7$$
* Para obtener $a_1$:
$$a_1 + 2d = a_3$$
$$a_1 = a_3 - 2d$$

```python
d=(1501-500)/7
a1 = 500 - 2*d
def a(n):
    return a1 + (n-1)*d
print(f'1) La app comenzó con {a(1)} usuarios')
print(f'2) El día 31 de enero habrá {a(31)} usuarios')

#Forma 1 => Tanteo
for i in range (32,40):
    print(f'Día {i} = {a(i)} ')
print(f'3) Hay 5505 usuarios luego de 1 mes y 7 días.')
```

```python
valor = 5505
dif = valor
i=0
while dif > 0:
    i+=1
    dif = valor - a(i)
if dif ==0:
    print(f'El término pertenece y ocupa la posición {i}')
else: 
    print('El término NO pertenece a la sucesión')
```

# Problema 12

Una empresa de publicidad ofrece para un puesto de trabajo un salario inicial y un aumento mensual en una razón constante. Javier se presenta a la entrevista, y el encargado del departamento de recursos humanos le explica que en el segundo mes de trabajo su sueldo será de $\$402.000$, y que en el tercer mes de trabajo será de $\$404.010$. Determine:

1. El salario aproximado que Javier recibiría en el décimo mes de trabajo en la empresa.
2. ¿Cuál sería el ingreso total recibido por los sueldos de Javier durante el tercer año de trabajo?

* Como indica que hay un aumento a "razon" constante => Significa que es una sucesión geométrica.

```python
a2 = 402000
a3 = 404010

r = a3 / a2
a1 = a2 / r

def a(n):
    return a1 * r ** (n-1)

print(f'El salario de javier al décimo mes de trabajo será de ${round(a(10))} pesos.')

suma = 0

for i in range(25,37):
    suma += a(i)
print(f'El ingreso total recibido por sueldos de javier durante el tercer año de trabajo será ${round(suma)} pesos.')
```

# Problema 13

Una empresa de ciberseguridad capacita trabajadores cada trimestre. En el primer trimestre capacitaron a $30$ trabajadores, en el segundo trimestre a $60$ trabajadores, y en el tercer trimestre a $120$ trabajadores. Si en los siguientes trimestres la empresa mantiene el mismo patrón de crecimiento, determine utilizando *Python*:

1. ¿A cuántos trabajadores capacitarán en el octavo trimestre?
2. Si la empresa lleva en el mercado $3$ años, ¿a cuántos trabajadores han capacitado en total?

* Es una sucesión geométrica, porque va aumentando a razón constante $r=2$

```python
a1 = 30

r = 2
def a(n):
    return a1 * r **(n-1)

print(f'En el octavo trimestre se capacitan {a(8)} trabajadores.')
suma = 0
for i in range(1,13):
    suma += a(i)
print(f'El total de trabajadores capacitados en 3 años es de  {suma}.')
```

Una empresa desarrolladora de videojuegos está monitoreando el crecimiento de jugadores activos diarios. Se sabe que el día 3 se incorporaron 1400 jugadores, y el día 8 se incorporaron 2900 jugadores. Si la cantidad de jugadores nuevos que se incorporan al juego crecen a partir de un valor constante día a día. 

1) Determine la cantidad de jugadores nuevos el día 10
2) Determine el total de jugadores que ha jugado el videojuego luego de 12 días. 

* Sucesión aritmética: Se incorporan diariamente una cantidad constante de jugadores. 
$$a_3 = 1400$$
$$a_8 = 2900$$
* Falta determinar $d$ y $a_1$

```python
a8 = 2900
a3 = 1400
d = (a8-a3)/5
a1 = a3 - 2*d
def a(n):
    return a1 + (n-1)*d
print(f'1) Los jugadores nuevos el décimo día son {a(10)} usuarios')

suma = 0
for i in range (1,13):
    suma += a(i)
print(f'2) La cantidad total de jugadores en 12 días es de {suma} usuarios')
```
